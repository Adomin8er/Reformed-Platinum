```python
import json
import pandas as pandas
```

```python
with open(r"/Users/adamholderness/Documents/Personal/Reformed-Platinum/RefPlat 1.1 Pokemon Dump.json", 'r') as refPlat_pokemonDump_json:
    refPlat_pokemonDump = json.load(refPlat_pokemonDump_json)

with open(r"/Users/adamholderness/Documents/Personal/Reformed-Platinum/RefPlat 1.1 Trainer Dump.json", 'r') as refPlat_trainerDump_json:
    refPlat_trainerDump = json.load(refPlat_trainerDump_json)
```

```python
trainer_numbers = [1, 2, 3]
trainer_instance = 0

# Build lookup once
pokemon_stats_lookup = {
    pokemon['readOnly']['name'].lower(): sum(pokemon['baseStats'].values())
    for pokemon in refPlat_pokemonDump
}

for trainer in refPlat_trainerDump:
    if trainer["readOnly"]['trainerID'] in trainer_numbers:
        party_list = "\n".join(
            f"{pokemon['species']} (Form: {pokemon['formID']}, Lv: {pokemon['level']}, Sex: {pokemon['sex']}, "
            f"Ability: {pokemon['ability']}, Base Stats: {pokemon_stats_lookup.get(pokemon['species'].lower(), 'N/A')})"
            for pokemon in trainer['party']
        )
        print(
f"""
======================
#{trainer['readOnly']['trainerID']} {trainer['readOnly']['name']}
======================
Trainer Instance - {trainer_instance}
----------------------
{party_list}
"""
        )
        trainer_instance += 1
```

```python

typing = 'normal'
baseStatTotal = 245

for pokemon in refPlat_pokemonDump:
    if (
        typing.lower() in [type.lower() for type in pokemon['typing']]
        and (baseStatTotal-100) <= sum(pokemon['baseStats'].values()) <= baseStatTotal+100 
        and pokemon['readOnly']['dexID'] <= 493
    ):
        print(f"#{pokemon['readOnly']['dexID']} {pokemon['readOnly']['name']} - {sum(pokemon['baseStats'].values())}")
```

```python
trainer_pokemon = 'Wooper'
pokemon_level = 7

for pokemon in refPlat_pokemonDump:
    if pokemon['readOnly']['name'].lower() == trainer_pokemon.lower():
        # Filter moves that can be learned at or below the given level
        level_up_moves = [
            move['move'] for move in pokemon['levelUpMoves']
            if move['level'] <= pokemon_level
        ]

        print(
f"""
#{pokemon['readOnly']['dexID']} {pokemon['readOnly']['name']}
-------------
Form - {pokemon['readOnly']['formIndex']}
Typing - {pokemon['typing']}
Stats - {sum(pokemon['baseStats'].values())}, {pokemon['baseStats']}
Abilities - {pokemon['abilities']}
Level {pokemon_level} Moves - {level_up_moves}
Egg Moves - {pokemon['eggMoves']}
TM Moves - {pokemon['compatibleTMs']}
"""
        )
```
